# 基底関数のグラフ

```python
import numpy
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

L = 10.0
N = 24
n = 4

xa = numpy.linspace(0, L, N + 1)

fig = plt.figure(figsize=(10,4))
ax1 = fig.add_subplot(1,2,1, projection='3d')
ax1.set_ylim(0, 2*n)

for j in range(n):
    ja = numpy.full(N + 1, j)
    za = numpy.full(N + 1, 0)
    za[j] = 1.0
    ax1.plot(xa, ja, za, marker='o')

ax1.set_xlabel('z')
ax1.set_ylabel('j')
ax1.set_zlabel('x')
ax1.set_title('z=P_j(x)')

ax2 = fig.add_subplot(1,2,2, projection='3d')
ax2.set_ylim(0, 2*n)
# Plotting planar wave basis function W_l(x)= exp(i*k_l*x)
for l in range(n):
    la = numpy.full(N + 1, l)
    kl = 2.0 * numpy.pi * l / L
    wa = numpy.exp(1j * kl * xa)
    ax2.plot(xa, la, wa.real, marker='o', label='Re|W_{}(x)|'.format(l))

ax2.set_xlabel('x')
ax2.set_ylabel('l')
ax2.set_zlabel('Re|W_l(x)|')
ax2.set_title('W_l(x) = exp(i*k_l*x)')
plt.show()
```

```python
# 2D plots of P_j(x) and W_l(x)
import numpy as np
import matplotlib.pyplot as plt

# Reuse parameters to match the 3D plots
L = 10.0
N = 24
n = 4
xa = np.linspace(0, L, N + 1)

fig, (ax_left, ax_right) = plt.subplots(1, 2, figsize=(8, 3), sharex=True)

# Left: P_j(x) plotted for j = 0..n-1
# P_j(x) is 1 at x_j and 0 elsewhere, with x_j taken at the j-th grid point
for j in range(n):
    # Construct P_j over the same grid as in the 3D plot
    Pj = np.zeros(N + 1)
    Pj[j] = 1.0
    ax_left.plot(xa, Pj, marker='o', label=f'P_{j}(x)')

ax_left.set_title('P_j(x) for j=0..{n-1}')
ax_left.set_xlabel('x')
ax_left.set_ylabel('P_j(x)')
ax_left.set_xlim(0, L)
ax_left.set_ylim(-0.1, 1.1)
ax_left.legend(loc='upper right', fontsize=8)

# Right: Re(W_l(x)) = cos(k_l x) for l = 0..n-1
for l in range(n):
    kl = 2.0 * np.pi * l / L
    wl = np.cos(kl * xa)  # real part of exp(i*k_l*x)
    ax_right.plot(xa, wl, marker='o', label=f'Re W_{l}(x)')

ax_right.set_title('Re W_l(x) = cos(k_l x)')
ax_right.set_xlabel('x')
ax_right.set_ylabel('Re(W_l(x))')
ax_right.set_xlim(0, L)
ax_right.set_ylim(-1.1, 1.1)
ax_right.legend(loc='upper right', fontsize=8)

plt.tight_layout()
plt.show()
```
